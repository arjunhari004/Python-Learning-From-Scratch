```python
#Create a list within func and pass the values to append it to the list

def values(a, b, c):
    list = []
    list.append(a)
    list.append(b)
    list.append(c)
    print(list)

values(10, 20, 30)
```

```python
#Create a func called d1()and print numbers 1-5 using while-loop

def d1():
    i = 1
    while i <= 5:
        print(i)
        i += 1
d1()
```

```python
#Create a func called maths() pass the values and perform add,sub,mul,div,mod,power

def maths(a, b):
    print("Addition:",a + b)
    print("Subtraction:",a - b)
    print("Multiplication:",a * b)
    print("Division:",a / b)
    print("Modulus:",a % b)
    print("Power:",a ** b)
maths(10, 3)   
```
