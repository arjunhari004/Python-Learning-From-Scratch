```python
#9. Print numbers from 1 to 5 and stop when number becomes 5

# Using while loop
i = 1
while(i <= 5):
    print(i)
    if(i == 5):
        break
    i = i + 1

# Using recursion
def numbers(i):
    print(i)
    if(i == 5):
        return
    numbers(i + 1)

numbers(1)
```
