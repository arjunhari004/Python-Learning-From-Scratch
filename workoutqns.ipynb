```python
#6.First Non Repeating char in swssip.
text = "swssip"
for i in range(len(text)):
    count = 0
    for j in range(len(text)):
        if text[i] == text[j]:
            count = count + 1
    if count == 1:
        print("First non-repeating character:", text[i])
        break
```

```python
#5.String Expansion a3b4c1

text = "a3b4c1"
result = ""
i = 0
while i < len(text):
    letter = text[i]
    number = int(text[i + 1])
    result = result + letter * number
    i = i + 2
print("Expanded string:", result)
```

```python
#4.Maximum difference [7,4,1,2,3] mac_d=7-1-->6

numbers = [7, 4, 1, 2, 3]

maximum = numbers[0]
minimum = numbers[0]

for number in numbers:

    if number > maximum:
        maximum = number

    if number < minimum:
        minimum = number

difference = maximum - minimum

print("Maximum:", maximum)
print("Minimum:", minimum)
print("Maximum Difference:", difference)
```

```python
#3.Merge 2 list without methods/function and arrange it in descending order

list1 = [10, 30, 20]
list2 = [50, 40, 60]
new_list = []
for i in list1:
    new_list.append(i)
for i in list2:
    new_list.append(i)
for i in range(len(new_list)):
    for j in range(i + 1, len(new_list)):
        if new_list[i] < new_list[j]:
            temp = new_list[i]
            new_list[i] = new_list[j]
            new_list[j] = temp

print("First list:", list1)
print("Second list:", list2)
print("Merged list in descending order:", new_list)
```

```python
#Create a list of students as well as their marks find the lowest as well as the highest marks print it along with their names

name = ["Arjun", "Rahul", "Akhil", "Vishnu", "Amal"]
marks = [92, 98, 98, 65, 78]

highest = marks[0]
lowest = marks[0]

for mark in marks:
    if mark > highest:
        highest = mark

    if mark < lowest:
        lowest = mark

print("Highest Marks:", highest)

for i in range(len(marks)):
    if marks[i] == highest:
        print("Student:", name[i])

print("Lowest Marks:", lowest)

for i in range(len(marks)):
    if marks[i] == lowest:
        print("Student:", name[i])
```

```python
#create a list full of palindrome and the find the shortest as well as longest palindrome in the list

numbers = [121, 11, 444, 12321, 22, 55555, 1001]

shortest = numbers[0]
longest = numbers[0]
for num in numbers:
    if len(str(num)) < len(str(shortest)):
        shortest = num
    if len(str(num)) > len(str(longest)):
        longest = num
print("List:", numbers)
print("Shortest palindrome:", shortest)
print("Longest palindrome:", longest)
```

```python
#create a list and count the number of palindromes

numbers = [121, 123, 444, 567, 101, 222]
count = 0
for num in numbers:
    original = num
    rev = 0
    while num > 0:
        digit = num % 10
        rev = rev * 10 + digit
        num = num // 10
    if original == rev:
        count += 1
print("Number of palindrome numbers:", count)
```
