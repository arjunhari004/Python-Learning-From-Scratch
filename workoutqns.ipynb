```python
students = [
    ["Arjun", 92],
    ["Abhi", 98],
    ["Ashwin", 98],
    ["Abin", 65],
    ["Amal", 65]
]

highest = students[0][1]
lowest = students[0][1]

for student in students:
    if student[1] > highest:
        highest = student[1]

    if student[1] < lowest:
        lowest = student[1]

print("Highest Marks:", highest)
print("Students:")

for student in students:
    if student[1] == highest:
        print(student[0])

print("Lowest Marks:", lowest)
print("Students:")

for student in students:
    if student[1] == lowest:
        print(student[0])
```

```python
#create a list full of palindrome and the find the shortest as well as longest palindrome in the list

numbers = [121, 11, 444, 12321, 22, 55555, 1001]

shortest = numbers[0]
longest = numbers[0]
for num in numbers:
    if len(str(num)) < len(str(shortest)):
        shortest = num
    if len(str(num)) > len(str(longest)):
        longest = num
print("List:", numbers)
print("Shortest palindrome:", shortest)
print("Longest palindrome:", longest)
```

```python
#create a list and count the number of palindromes

numbers = [121, 123, 444, 567, 101, 222]
count = 0
for num in numbers:
    original = num
    rev = 0
    while num > 0:
        digit = num % 10
        rev = rev * 10 + digit
        num = num // 10
    if original == rev:
        count += 1
print("Number of palindrome numbers:", count)
```
