```python
#create a list of names and count how many times a is repeated 

names = ["Arjun", "Rahul", "Akhil", "Anand", "Amal"]

count = 0

for name in names:
    for letter in name:
        if letter == "a" or letter == "A":
            count = count + 1

print("The letter a is repeated", count, "times")
```

```python
#create a nexted list with full of strings and check whether palindrome or not

names = [["madam", "hello"], ["malayalam", "world"], ["level", "python"]]

for group in names:
    for word in group:

        if word == word[::-1]:
            print(word, "is Palindrome")
        else:
            print(word, "is Not Palindrome")
```

```python
#without bultin methods

names = eval(input("Enter names: "))
marks = eval(input("Enter marks: "))

highest = marks[0]
highest_name = names[0]

for i in range(len(marks)):

    if marks[i] > highest:
        highest = marks[i]
        highest_name = names[i]

longest = names[0]

for name in names:

    if len(name) > len(longest):
        longest = name

print("Highest mark:", highest)
print("Student name:", highest_name)

print("Longest name:", longest)
```

```python
#Create a list of students as well as their marks find the lowest as well as the highest marks print it along with their names

name = ["Arjun", "Rahul", "Akhil", "Vishnu", "Amal"]
marks = [92, 98, 98, 65, 78]

highest = marks[0]
lowest = marks[0]

for mark in marks:
    if mark > highest:
        highest = mark

    if mark < lowest:
        lowest = mark

print("Highest Marks:", highest)

for i in range(len(marks)):
    if marks[i] == highest:
        print("Student:", name[i])

print("Lowest Marks:", lowest)

for i in range(len(marks)):
    if marks[i] == lowest:
        print("Student:", name[i])
```

```python
#create a list full of palindrome and the find the shortest as well as longest palindrome in the list

numbers = [121, 11, 444, 12321, 22, 55555, 1001]

shortest = numbers[0]
longest = numbers[0]
for num in numbers:
    if len(str(num)) < len(str(shortest)):
        shortest = num
    if len(str(num)) > len(str(longest)):
        longest = num
print("List:", numbers)
print("Shortest palindrome:", shortest)
print("Longest palindrome:", longest)
```

```python
#create a list and count the number of palindromes

numbers = [121, 123, 444, 567, 101, 222]
count = 0
for num in numbers:
    original = num
    rev = 0
    while num > 0:
        digit = num % 10
        rev = rev * 10 + digit
        num = num // 10
    if original == rev:
        count += 1
print("Number of palindrome numbers:", count)
```
