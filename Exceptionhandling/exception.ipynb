```python
# 8.Write a program to handle multiple exception using a single except block

try:
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))
    result = a / b
    print("Result:", result)
except (ValueError, ZeroDivisionError):
    print("Invalid input or Cannot divide by zero")
finally:
    print("Thank You")
```

```python
# 9.Create a nested try-except program.

try:
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))
    try:
        result = a/b
        print("Result:", result)
    except ZeroDivisionError:
        print("Cannot divide by zero")
except ValueError:
    print("Invalid Input")
finally:
    print("Thank You")
```

```python
# 10.Create a user-defined exception for checking student marks.

class InvalidMarks(Exception):
    def __init__(self):
        print("Invalid Marks Exception")
try:
    marks = int(input("Enter student marks: "))
    if marks<0 or marks>100:
        raise InvalidMarks
    else:
        print("Student Marks:",marks)
        print("Valid Marks")
except ValueError:
    print("Please enter numbers only")
except InvalidMarks:
    print("Marks should be between 0 and 100")
finally:
    print("Thank You")
```

```python
# 4. Find the sum of numbers from 1 to N 

n = int(input("Enter a number: "))
sum = 0

for i in range(1, n+1):
    sum = sum + i

print("Sum:", sum)
```

```python
# 5.Find the factorial of a number without using **

n = int(input("Enter a number: "))
fact = 1

for i in range(1, n+1):
    fact = fact * i

print("Factorial:", fact)
```

```python
# 6.count the number of digits in a number

n = int(input("Enter a number: "))
count = 0
while n > 0:
    n = n // 10
    count = count + 1
print("Number of digits:", count)
```
