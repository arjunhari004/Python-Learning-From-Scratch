```python
try:
    a = int(input("Enter A number:"))
    b = int(input("Enter another number: "))
    print(a*b)

# try block

except ZeroDivisionError:
    print("Cannot divide By zero")
# except value error
except ValueError:
    print("please enter number only")
```

```python
try:
    a = int(input("Enter a number:"))
    b = int(input("Enter another number"))
    result = a/b
except ValueError:
    print("Invalid Input")
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Result:",result)
finally:
    print("Thank You")                
```
