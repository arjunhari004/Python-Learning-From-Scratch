```python
# Q4. Change Student Details
# Create a class Student with a constructor that accepts name and age.
# Create a method update_details() that changes the student's name and age. Create another method to display the updated details.
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def update_details(self):
        self.name = input("Enter name")
        self.age = int(input("Enter new age"))
    def display(self):
        print(self.name)
        print(self.age)

s1 = Student("Arjun",22)

print("Original details")
s1.display()

print("Updating details...")
s1.update_details()

print("Updated details..")
s1.display()
```

```python
# Q6. Rectangle Calculation
# Create a class Rectangle with a constructor that accepts length and breadth. Create methods to calculate and display the area and perimeter.
class rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def calculator(self):
        area = self.length * self.breadth
        perimeter = 2 * (self.length + self.breadth)

        print("Area:", area)
        print("Perimeter:", perimeter)

r1=rectangle(10,5)
r1.calculator()
```

```python
# Q7. Circle Calculation
# Create a class Circle with a constructor that accepts radius. Create methods to calculate the area and circumference.
class circle:
    def __init__(self,radius):
        self.radius=radius
    def calculate(self):
        area=3.14*self.radius*self.radius
        circumference=2*3.14*self.radius

        print("area of circle is :",area)
        print("circumference of circle is:",circumference)

c1=circle(3)
c1.calculate()
```

```python
# Q8. Bank Account
# Create a class BankAccount with a constructor that accepts account_holder and balance. Create methods to:
# Display account details.
# Deposit an amount.
# Display the updated balance.

class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance

    def account_details(self):
        print("account holder :", self.account_holder)
        print("balance :", self.balance)

    def deposit(self,ammount):  
        self.balance = self.balance + ammount

    def display_balance(self):
        print("Updated Balance:", self.balance)
    
a1=BankAccount("abhinand",50000)

a1.account_details()
a1.deposit(500)
a1.display_balance()
```

```python
# Q9. Student Marks
# Create a class Student with a constructor that accepts name, mark1, mark2, and mark3. Create methods to calculate the total and average marks.
class student:
    def __init__(self,name,mark1,mark2,mark3):
        self.name=name
        self.mark1=mark1
        self.mark2=mark2
        self.mark3=mark3

    def total(self):
        total = self.mark1 + self.mark2 + self.mark3
        print("student total mark :", total)

    def average(self):
        avg = (self.mark1 + self.mark2 + self.mark3) / 3
        print("Average Marks:", avg)

s1=student("Abu",29,40,38)
s1.total()
s1.average()
```

```python
# Q10. Product Discount
# Create a class Product with a constructor that accepts name and price.
#  Create a method to calculate a 10% discount and another method to display the final price.
 
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price

    def calculator(self)  : 
        self.discount = self.price * 0.10
        self.final_price = self.price - self.discount

    def display(self)  :
        print("product name :",self.name)
        print("product price :",self.price)
        print("discount of product :", self.discount)
        print("final price ,", self.final_price)  

d1=product("samsung s26 ultra",1300000)
d1.calculator()
d1.display()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def compare(self, p):
        if self.price > p.price:
            print(self.name, "is more expensive")
        elif self.price < p.price:
            print(p.name, "is more expensive")
        else:
            print("Both products have the same price")


p1 = Product("Phone", 20000)
p2 = Product("Laptop", 50000)

p1.compare(p2)
```
