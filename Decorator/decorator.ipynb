```python
def message(func): #func --> add
    def inner():
        print("Calculation started")
        func()  #add()
        print("Calculation completed")
    return inner

@message  #take msg func as decorator for this function add
def add():
    a = 10
    b = 20
    print("Sum:",a + b)   
add()     
```
